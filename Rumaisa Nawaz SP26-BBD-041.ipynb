```python
# Program to count word frequency
text= "Python is easy and Python is powerful and easy to learn"
words= text.lower().split()
frequency={}
for word in words:
    frequency[word]=frequency.get(word,0)+1
print("Word Frequency:")
for word, count in frequency.items():
    print(word, ":", count)
```

```python
#find most popular product
orders=['electronics', 'clothing','electronics', 'food', 'clothing', 'electronics']
count={}
for category in orders:
    count[category]= count.get(category,0)+1
most_popular=max(count, key=count.get)
print("Most Popular Category:", most_popular)
```

```python

# Inventory Management
inventory = {
    "Laptop": {"quantity": 3, "price": 100000},
    "Mouse": {"quantity": 10, "price": 2000},
    "Keyboard": {"quantity": 4, "price": 5000}
}
# Update quantity
inventory["Laptop"]["quantity"] = 2
# Calculate total value
total = 0
for item in inventory:
    total += inventory[item]["quantity"] * inventory[item]["price"]
print("Total Inventory Value:", total)
# Low stock items
print("Low Stock Items:")
for item in inventory:
    if inventory[item]["quantity"] < 5:
        print(item)
```

```python
# Student Gradebook
students={
    "Ali":{"math":80, "english":70},
    "Sara":{"math":90, "english":85},
    "Ahmad":{"math":60, "english":75}
}
#Average per student
print("Average Grades:")
for s in students:
    avg= sum(students[s].values())/len(students[s])
    print(s,":", avg)
math_marks={s: students[s]["math"] for s in students}
print("Highest in math:", max(math_marks,key=math_marks.get))
print("Lowest in math:", min(math_marks, key=math_marks.get))
```

```python
#Simple E-commerce store
products={
    "shirt":{"price":1500, "quantity":10},
    "shoes":{"price":4000, "quantity":4}
}
cart={}
#add to cart
cart["shirt"]=2
cart["shoes"]=3
total=0
for item in cart:
    total+=cart[item]*products[item]["price"]
print("Total Cost:", total)
total= total-(total*0.1) #10% discount
total= total+(total*0.05) #5% dicount
print("Final Cost:", total)
```
