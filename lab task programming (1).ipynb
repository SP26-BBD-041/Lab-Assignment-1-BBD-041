```python
from datetime import datetime  #question no 1
def greet_with_time(name):
    current_time=datetime.now().strftime("%H:%M:%S")
    print(f"Hello,{name}! current time is {current_time}")
greet_with_time("Faria")
```

```python
def calculate_average(num1,num2,num3): #question no 2
    average=(num1+num2+num3)/3
    return average
print("Average:",calculate_average(13,45,23))
```

```python
def calculate_discount(price,discount=10): #question no 3
    final_price=price-(price*discount/100)
    return final_price
print("Discounted price:",calculate_discount(10000))
print("Discounted price:",calculate_discount(10000,20))
```

```python
def print_info(name,age,city): #Question no 4
    print(f"Name: {name}")
    print(f"Age: {age}")
    print(f"City: {city}")
print_info(name="Zoha",age=21,city="Islamabad")
```

```python
def find_max(*numbers): #Question no 5
    return max(numbers)
print(find_max(23,56,87,21,90))
```

```python
def factorial(n): #Question no 6
    if n == 0 or n == 1:
        return 1
    return n*factorial(n-1)
print(factorial(6))
```

```python
def square(x): #Question no 7
    return x*x
def apply_function(func,numbers):
    result= []
    for num in numbers:
        result.append(func(num))
    return result 
print(apply_function(square,[1,2,3,4]))
```

```python
def add_five(x): #Question no 9
    return x+5
def apply_operation(func,numbers):
    result = []
    for num in numbers:
        result.append(func(num))
    return result
print(apply_operation(add_five,[1,2,3,4]))
```

```python
import time #Question no 10
def timer_decorator(func):
    def wrapper():
        start=time.time()
        func()
        end = time.time()
        print("Execution Time:",end-start)
    return wrapper
@timer_decorator
def greet():
    print("Hello!")
    time.sleep(2)
greet()
```
